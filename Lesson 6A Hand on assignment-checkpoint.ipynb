```python
#Hand on exercise2(banking & cybersecurity industry)

log_message=" LOGIN SUCCESSFUL FROM ATM-045 ON 2026-07-25 "
cleaned_log=log_message.strip()
lowercase=cleaned_log.lower()
successful_position=cleaned_log.find("SUCCESSFUL")
print("Log Message:",log_message)
print("Cleaned Log:",cleaned_log)
print("Lower case:",lowercase)
print("Postion of success:", successful_position)
print("Updated log:",cleaned_log.replace("SUCCESSFUL","VERIFIED"))
```
