```python
#Gift Task: Drug Inventory Management

#Task 1
drug_name="Paracetamol"
price=850
stock_quantity=250
quantity_sold=65
minimum_stock=100
expired=False 

#Task 2
remaining_stock = stock_quantity - quantity_sold
print(remaining_stock)

#Task 3
revenue=price*quantity_sold
print(revenue)

#Task 4
stock_quantity+=remaining_stock
print(stock_quantity)

#Task 5
restock=remaining_stock <= minimum_stock
print(restock)

#Task 6
print(expired)

#Task 7
sell_drug=stock_quantity>minimum_stock and not expired
print(sell_drug)

#Task 8
is_eligible=remaining_stock>200 or revenue>50000
print(is_eligible)

#Task 9
print("="*35)
print("DRUG INVENTORY REPORT")
print("="*35)
print("Drug Name:", drug_name)
print("Remaining Stock:", remaining_stock)
print("Revenue:", revenue)
print("Stock quantity:", stock_quantity)
print("Restock:", restock)
print("Expired:",expired)
print("Safe to sell:", sell_drug)
print("Eligible for Discount:", is_eligible)

print("="*35)
```
