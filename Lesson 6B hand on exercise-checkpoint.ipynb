```python
#Hand on Exercise 1(human resource)

employee_record="EMP-HR-2026-004512"

employee_prefix= employee_record[0:3]
city_code= employee_record[4:6]
registration_year= employee_record[7:11]
customer_number= employee_record[12:]
first_10_char= employee_record[0:9]
last_6_char= employee_record[-6:]
every_second_char= employee_record[::2]
every_third_char= employee_record[::3]
reverse_record= employee_record[::-1]

print("Employee prefix:",employee_prefix)
print("City Code:",city_code)
print("Registration Year:",registration_year)
print("Customer Name:",customer_number)
print("First 10 Char:",first_10_char)
print("Last 6 Char:",last_6_char)
print("Every Second Char:",every_second_char)
print("Every third Char:",every_third_char)
print("Reverse Record:",reverse_record)
```

```python
#Hand on exercise 2 (banking and finance)

card_number= "5489123476549876"
print("First 4 digits:",card_number[0:4])
print("Last 4 digits:",card_number[-4:])
print("Middle 8 digits:",card_number[4:12])
print("Every second digits:",card_number[::2])
print("Every third digit:",card_number[::3])
print("Reverse Card Number:",card_number[::-1])
print("Reverse last 8 digit:",card_number[-8:][::1])
```
